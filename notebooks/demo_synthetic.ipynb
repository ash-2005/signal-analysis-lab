This notebook demonstrates how lag distorts cross-correlation and how to recover it.

```python
import numpy as np
import matplotlib.pyplot as plt
from src.generator import generate_coupled_pair
from src.crosscorr import lagged_xcorr, peak_lag

x, y = generate_coupled_pair(1000, lag=0.4, noise_std=0.1, fs=100.0)
res = lagged_xcorr(x, y, max_lag=1.0, fs=100.0)
lag, _ = peak_lag(res)
print(f"Estimated peak lag: {lag:.3f}s (Ground truth: 0.4s)")
```
